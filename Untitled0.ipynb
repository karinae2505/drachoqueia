```python
def preguntar_agente(pregunta):
    prompt = f"""Sos un asistente legal llamado "Dra. Choque" que ayuda a personas
que sufrieron un accidente de tránsito, basándote ÚNICAMENTE en el siguiente documento.

DOCUMENTO:
{texto_documento}

Respondé la siguiente pregunta de forma clara y breve, basándote solo en la información
del documento. Si la pregunta no se puede responder con el documento, decilo honestamente.

PREGUNTA: {pregunta}

RESPUESTA:"""

    respuesta = modelo.generate_content(prompt)
    return respuesta.text

# Prueba
pregunta_prueba = "Tuve un choque y no tengo lesiones visibles, ¿igual debo hacer algo?"
print("PREGUNTA:", pregunta_prueba)
print("\nRESPUESTA:")
print(preguntar_agente(pregunta_prueba))
```

```python
preguntas = [
    "¿Cuánto tiempo tengo para reclamar después de un accidente?",
    "¿Qué pasa si el otro auto no tiene seguro?",
    "¿Puedo firmar un acuerdo económico en el momento del choque?",
    "¿Qué hago si el conductor se dio a la fuga?",
    "¿Cómo preparo una ensalada César?"
]

for p in preguntas:
    print("PREGUNTA:", p)
    print("RESPUESTA:", preguntar_agente(p))
    print("-" * 80)
```
